Model 1: Linear Regression Using All Features

```python
## Take every column except medv. ,, conventially the X variable is always caps.. Columns so its a dataframe. 
## since were using all columns just use existing df and drop thr target column
X = df.drop(columns=["medv"])
```

```python
## target vector,, conventially y is lower case .. its a series just the target column 
y = df["medv"]
```

```python
## the input dataframe were using as features, first 5 rows,, basically the same df but no medv
X.head()
```

```python
## first 5 rows of target, the prices (medv)
y.head()
```

```python
from sklearn.linear_model import LinearRegression    # Machine learning algorithm
```

```python
## creating algorithm no data yet 
model1 = LinearRegression()
```

```python
## Putting the features and target in the model (feeding it data).. makes line of best fit 
##,searching for coeficients for line that that produce the smallest total squared resisduals,, least error
##Calculate coefficient that minimize squared residuals
# the X is the features will always be df, the y is the target and is gonna be 1 column (series MEDV)

model1.fit(X, y)
```

```python
## Evaluate model r2 value for model1, how accurate does all features predict house prices.. how good is the prediction?
## use the line, make predictions, compare with actual target and then see how will did those features predict.. 
# Its the model test score

## While the line of best fit is the prediction equation learned from the data, 
# while R² is an evaluation metric that measures how well that learned line explains the variation in the target values.

r2_model1 = model1.score(X, y)

print("Model 1 R²:", r2_model1)
```

```python
## make predictions 
predictions = model1.predict(X)


##first 5 predictions 
predictions[:5]
```

```python
##compare to actual targets 
y.head()
```

```python
##comparison df

comparison = pd.DataFrame({
    "Actual": y,
    "Predicted": predictions
})

comparison.head()
```

MODEL 2:: Linear Regression Using Only Average Number of Rooms

```python
## the feature now is just rooms from the df , input!
X_room = df[["rm"]]
```

```python
## call the algo
model2=LinearRegression()
```

```python
## feed the model the inputs which is df[["rm"]] ,(rooms), stored in X_room. 
# And target which is df["medv"], series, and the medv (prices) stored in y
## now the model will look for the line of best fit 
# by looking for the coeficent that get the least total squared residuals (least differences from the actual and predicted)
## now it saves the equation of the line for its predictions

model2.fit(X_room,y)
```

```python
## R2 VALUE FOR MODEL2

r2_model2 = model2.score(X_room,y)
print("model2 r^2 is", r2_model2)
```

```python
## predictions for df["medv"] using the line of best fit and the input of df[["rm"]] stored in X_room

predictions2 = model2.predict(X_room)

predictions2[:5]
```

# Model Comparison

### Model 1 (All 13 Features)
- Used all available housing features.
- R² = 0.733
- Explains about 73% of the variation in house prices.
- Produced the stronger model.

### Model 2 (Rooms Only)
- Used only the average number of rooms (rm).
- R² = 0.476
- Explains about 48% of the variation in house prices.
- Simpler model, but less accurate.

### Conclusion

The model using all 13 housing features performed better than the model using only the number of rooms. Even though the average number of rooms is one of the strongest predictors of house price (correlation with medv was .690), it does not capture all of the information contained in the dataset. Other features such as crime rate, property taxes, income level, and pupilteacher ratio also contribute to predicting house values, resulting in a higher R² score for Model 1.
